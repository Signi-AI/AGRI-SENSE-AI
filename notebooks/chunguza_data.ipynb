```python
import pandas as pd

df = pd.read_csv("../data/crop_recommendation.csv")

print("Umbo la data:", df.shape)
print("\nMajina ya columns:", list(df.columns))
print("\nMifano ya rows 5 za kwanza:")
df.head()
```

```python
from sklearn.model_selection import train_test_split

# X = vigezo tunavyotumia kutabiri (features)
X = df[['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']]

# y = kile tunachotaka kutabiri (target) - jina la zao
y = df['label']

# Gawa data: 80% kufundisha, 20% kupima
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.3, 
    random_state=42,
    stratify=y
)

# Thibitisha ukubwa wa kila sehemu
print("Ukubwa wa data ya kufundisha (X_train):", X_train.shape)
print("Ukubwa wa data ya kupima (X_test):", X_test.shape)
print("\nMgawanyo wa mazao kwenye data ya kufundisha:")
print(y_train.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Unda model - random_state inahakikisha matokeo yanarudiwa sawa kila wakati
model = RandomForestClassifier(random_state=42)

# Fundisha model kwa kutumia data ya training
model.fit(X_train, y_train)

# Tumia model kutabiri kwenye data ya test (ambayo haijawahi kuiona)
y_pred = model.predict(X_test)

# Pima usahihi
accuracy = accuracy_score(y_test, y_pred)
print(f"Usahihi wa model (accuracy): {accuracy:.4f}")
print(f"Kwa asilimia: {accuracy*100:.2f}%")
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import pandas as pd

# Ripoti ya kina - inaonyesha utendaji wa model kwa KILA zao kivyake
print("RIPOTI YA KINA (kwa kila zao):")
print(classification_report(y_test, y_pred))

# Confusion Matrix - inaonyesha ni zao gani linachanganywa na zao gani
cm = confusion_matrix(y_test, y_pred, labels=model.classes_)
cm_df = pd.DataFrame(cm, index=model.classes_, columns=model.classes_)

# Onyesha tu mazao yaliyokosewa (siyo yote 22x22, ili iwe rahisi kusoma)
print("\nMazao yenye makosa (kama yapo):")
for zao in model.classes_:
    makosa = cm_df.loc[zao].drop(zao).sum()  # jumla ya makosa kwa zao hili
    if makosa > 0:
        print(f"  {zao}: makosa {makosa}")
```

```python
def angalia_hali_ya_udongo(zao, usomaji_wa_sasa, jedwali_la_viwango):
    """
    zao: jina la zao (string), mfano 'maize'
    usomaji_wa_sasa: dictionary yenye vigezo 6, mfano:
        {'N': 40, 'P': 30, 'K': 20, 'temperature': 30, 'humidity': 65, 'ph': 5.0}
    jedwali_la_viwango: jedwali letu la min/max/mean kwa kila zao
    
    Inarudisha: orodha (list) ya matatizo yaliyogundulika
    """
    matatizo = []
    
    # Angalia kama zao lipo kwenye jedwali letu
    if zao not in jedwali_la_viwango.index:
        return [f"Samahani, zao '{zao}' halipo kwenye database yetu."]
    
    # Pitia kila kigezo kimoja baada ya kingine
    for kigezo in ['N', 'P', 'K', 'temperature', 'humidity', 'ph']:
        thamani_ya_sasa = usomaji_wa_sasa[kigezo]
        kiwango_cha_chini = jedwali_la_viwango.loc[zao, (kigezo, 'min')]
        kiwango_cha_juu = jedwali_la_viwango.loc[zao, (kigezo, 'max')]
        
        if thamani_ya_sasa < kiwango_cha_chini:
            tofauti = kiwango_cha_chini - thamani_ya_sasa
            matatizo.append({
                'kigezo': kigezo,
                'hali': 'chini_ya_kiwango',
                'thamani_ya_sasa': thamani_ya_sasa,
                'kiwango_kinachohitajika': kiwango_cha_chini,
                'tofauti': round(tofauti, 2)
            })
        elif thamani_ya_sasa > kiwango_cha_juu:
            tofauti = thamani_ya_sasa - kiwango_cha_juu
            matatizo.append({
                'kigezo': kigezo,
                'hali': 'juu_ya_kiwango',
                'thamani_ya_sasa': thamani_ya_sasa,
                'kiwango_kinachohitajika': kiwango_cha_juu,
                'tofauti': round(tofauti, 2)
            })
    
    return matatizo
```

```python
def hesabu_marekebisho_ya_ph(ph_ya_sasa, ph_lengwa, aina_ya_udongo='tifutifu'):
    """
    ph_ya_sasa: pH ya sasa ya shamba (kutoka sensor)
    ph_lengwa: pH inayohitajika (tunatumia 'mean' ya zao kutoka jedwali letu)
    aina_ya_udongo: 'mchanga', 'tifutifu', au 'mfinyanzi'
    
    Inarudisha: dictionary yenye aina ya marekebisho na kiasi
    """
    # Buffer factor (K) kwa kila aina ya udongo - tani za CaCO3 kwa hekta, kwa unit moja ya pH
    buffer_factor = {
        'mchanga': 1.75,      # wastani wa 1.5-2.0
        'tifutifu': 3.5,      # wastani wa 3.0-4.0
        'mfinyanzi': 6.5      # wastani wa 5.0-8.0
    }
    
    # Sulfur factor (S) kwa kila aina ya udongo - kg za sulfur kwa hekta, kwa unit moja ya pH
    sulfur_factor = {
        'mchanga': 280,
        'tifutifu': 420,
        'mfinyanzi': 560
    }
    
    tofauti_ya_ph = round(ph_lengwa - ph_ya_sasa, 2)
    
    if abs(tofauti_ya_ph) < 0.1:
        # Tofauti ndogo mno, hakuna hatua muhimu ya kuchukua
        return {
            'hatua_inahitajika': False,
            'ujumbe': 'pH iko karibu kabisa na kiwango kinachohitajika.'
        }
    
    if tofauti_ya_ph > 0:
        # pH ya sasa ni CHINI ya lengwa -> udongo una asidi nyingi -> ongeza CHOKAA
        kiasi = round(tofauti_ya_ph * buffer_factor[aina_ya_udongo], 2)
        return {
            'hatua_inahitajika': True,
            'aina_ya_marekebisho': 'chokaa (agricultural lime)',
            'kiasi': kiasi,
            'kipimo': 'tani kwa hekta',
            'sababu': 'Udongo una asidi nyingi kuliko inavyohitajika kwa zao hili.'
        }
    else:
        # pH ya sasa ni JUU ya lengwa -> udongo una alkali nyingi -> ongeza SULFURI
        kiasi = round(abs(tofauti_ya_ph) * sulfur_factor[aina_ya_udongo], 2)
        return {
            'hatua_inahitajika': True,
            'aina_ya_marekebisho': 'sulfuri (elemental sulfur)',
            'kiasi': kiasi,
            'kipimo': 'kg kwa hekta',
            'sababu': 'Udongo una alkali nyingi kuliko inavyohitajika kwa zao hili.'
        }
```

```python
# Tumia mfano wetu wa awali: maize, ph_ya_sasa = 5.0
ph_lengwa_ya_maize = viwango_vya_mazao.loc['maize', ('ph', 'mean')]
print("pH lengwa ya maize (mean):", round(ph_lengwa_ya_maize, 2))

matokeo_ya_ph = hesabu_marekebisho_ya_ph(
    ph_ya_sasa=5.0, 
    ph_lengwa=ph_lengwa_ya_maize, 
    aina_ya_udongo='tifutifu'
)
print(matokeo_ya_ph)
```

```python
def hesabu_marekebisho_ya_npk(kigezo, thamani_ya_sasa, thamani_lengwa):
    """
    kigezo: 'N', 'P', au 'K'
    thamani_ya_sasa: usomaji wa sasa kutoka sensor
    thamani_lengwa: kiwango kinachohitajika (mean kutoka jedwali)
    
    Inarudisha: dictionary yenye kiasi cha mbolea kinachohitajika
    """
    factor = {
        'N': 2.5,
        'P': 2.5,
        'K': 2.0
    }
    
    majina_kamili = {
        'N': 'Nitrogen',
        'P': 'Phosphorus',
        'K': 'Potassium'
    }
    
    tofauti = round(thamani_lengwa - thamani_ya_sasa, 2)
    
    if abs(tofauti) < 2:
        return {
            'hatua_inahitajika': False,
            'ujumbe': f'{majina_kamili[kigezo]} iko karibu na kiwango kinachohitajika.'
        }
    
    if tofauti > 0:
        kiasi = round(tofauti * factor[kigezo], 2)
        return {
            'hatua_inahitajika': True,
            'kigezo': majina_kamili[kigezo],
            'hali': 'upungufu',
            'kiasi': kiasi,
            'kipimo': 'kg kwa hekta (nutrient safi)',
            'sababu': f'{majina_kamili[kigezo]} imepungua kuliko inavyohitajika kwa zao hili.'
        }
    else:
        kiasi = round(tofauti * factor[kigezo], 2)
        return {
            'hatua_inahitajika': True,
            'kigezo': majina_kamili[kigezo],
            'hali': 'uzidi',
            'kiasi': abs(kiasi),
            'kipimo': 'kg kwa hekta (punguza kwa kiasi hiki)',
            'sababu': f'{majina_kamili[kigezo]} imezidi kuliko inavyohitajika kwa zao hili.'
        }
```

```python
def toa_ripoti_kamili(zao, usomaji_wa_sasa, jedwali_la_viwango, aina_ya_udongo='tifutifu'):
    """
    Function kuu inayounganisha utambuzi wa matatizo + mahesabu ya marekebisho.
    
    zao: jina la zao, mfano 'maize'
    usomaji_wa_sasa: dictionary ya N,P,K,temperature,humidity,ph
    jedwali_la_viwango: jedwali letu la min/max/mean
    aina_ya_udongo: kwa ajili ya formula ya pH
    
    Inarudisha: dictionary kamili yenye ripoti ya mkulima
    """
    # Hatua 1: Angalia zao lipo kwenye database
    if zao not in jedwali_la_viwango.index:
        return {'hitilafu': f"Zao '{zao}' halipo kwenye database yetu."}
    
    # Hatua 2: Gundua ni vigezo gani vyenye tatizo (tunatumia function ya B2)
    matatizo = angalia_hali_ya_udongo(zao, usomaji_wa_sasa, jedwali_la_viwango)
    
    # Hatua 3: Kama hakuna tatizo lolote
    if len(matatizo) == 0:
        return {
            'zao': zao,
            'hali_ya_jumla': 'Nzuri',
            'ujumbe': f'Hali ya udongo wako inafaa kwa {zao}. Hakuna marekebisho yanayohitajika.',
            'marekebisho': []
        }
    
    # Hatua 4: Kwa kila tatizo lililogundulika, hesabu marekebisho yanayofaa
    marekebisho_yote = []
    
    for tatizo in matatizo:
        kigezo = tatizo['kigezo']
        
        if kigezo == 'ph':
            ph_lengwa = jedwali_la_viwango.loc[zao, ('ph', 'mean')]
            marekebisho = hesabu_marekebisho_ya_ph(
                ph_ya_sasa=usomaji_wa_sasa['ph'],
                ph_lengwa=ph_lengwa,
                aina_ya_udongo=aina_ya_udongo
            )
            marekebisho['kigezo'] = 'pH'
            marekebisho_yote.append(marekebisho)
        
        elif kigezo in ['N', 'P', 'K']:
            thamani_lengwa = jedwali_la_viwango.loc[zao, (kigezo, 'mean')]
            marekebisho = hesabu_marekebisho_ya_npk(
                kigezo=kigezo,
                thamani_ya_sasa=usomaji_wa_sasa[kigezo],
                thamani_lengwa=thamani_lengwa
            )
            marekebisho_yote.append(marekebisho)
        
        else:
            # kwa temperature/humidity - hatuna formula ya kurekebisha (havidhibitiwi na mkulima moja kwa moja)
            marekebisho_yote.append({
                'kigezo': kigezo,
                'hatua_inahitajika': None,
                'ujumbe': f'{kigezo} iko nje ya wigo unaofaa, lakini haiwezi kurekebishwa moja kwa moja shambani (inategemea hali ya hewa).'
            })
    
    return {
        'zao': zao,
        'hali_ya_jumla': 'Kuna matatizo yanayohitaji hatua',
        'idadi_ya_matatizo': len(matatizo),
        'marekebisho': marekebisho_yote
    }
```

```python
def angalia_ph_ya_maji(ph_ya_maji, wigo_wa_chini=6.0, wigo_wa_juu=7.5):
    """
    ph_ya_maji: pH ya sasa ya maji ya umwagiliaji (kutoka sensor)
    wigo_wa_chini/wigo_wa_juu: kiwango cha jumla kinachokubalika kwa mazao mengi
    
    Inarudisha: dictionary yenye tathmini ya pH ya maji
    """
    if ph_ya_maji < wigo_wa_chini:
        return {
            'kigezo': 'pH ya Maji',
            'hatua_inahitajika': True,
            'hali': 'asidi_nyingi',
            'thamani_ya_sasa': ph_ya_maji,
            'wigo_unaofaa': f'{wigo_wa_chini}-{wigo_wa_juu}',
            'ujumbe': 'Maji ya umwagiliaji yana asidi nyingi. Hii inaweza kufanya virutubisho fulani (kama Calcium na Magnesium) visipatikane vizuri kwa mmea.'
        }
    elif ph_ya_maji > wigo_wa_juu:
        return {
            'kigezo': 'pH ya Maji',
            'hatua_inahitajika': True,
            'hali': 'alkali_nyingi',
            'thamani_ya_sasa': ph_ya_maji,
            'wigo_unaofaa': f'{wigo_wa_chini}-{wigo_wa_juu}',
            'ujumbe': 'Maji ya umwagiliaji yana alkali nyingi. Hii inaweza kufanya Iron, Manganese, na Zinc visipatikane vizuri kwa mmea, hata kama vipo kwenye udongo.'
        }
    else:
        return {
            'kigezo': 'pH ya Maji',
            'hatua_inahitajika': False,
            'ujumbe': 'pH ya maji ya umwagiliaji iko ndani ya wigo unaofaa.'
        }
```

```python
def toa_ripoti_kamili(zao, usomaji_wa_sasa, jedwali_la_viwango, aina_ya_udongo='tifutifu', ph_ya_maji=None):
    """
    Function kuu inayounganisha utambuzi wa matatizo + mahesabu ya marekebisho.
    
    zao: jina la zao, mfano 'maize'
    usomaji_wa_sasa: dictionary ya N,P,K,temperature,humidity,ph (ya UDONGO)
    jedwali_la_viwango: jedwali letu la min/max/mean
    aina_ya_udongo: kwa ajili ya formula ya pH ya udongo
    ph_ya_maji: (HIARI) pH ya maji ya umwagiliaji, kama mkulima anayo
    
    Inarudisha: dictionary kamili yenye ripoti ya mkulima
    """
    if zao not in jedwali_la_viwango.index:
        return {'hitilafu': f"Zao '{zao}' halipo kwenye database yetu."}
    
    matatizo = angalia_hali_ya_udongo(zao, usomaji_wa_sasa, jedwali_la_viwango)
    
    marekebisho_yote = []
    
    for tatizo in matatizo:
        kigezo = tatizo['kigezo']
        
        if kigezo == 'ph':
            ph_lengwa = jedwali_la_viwango.loc[zao, ('ph', 'mean')]
            marekebisho = hesabu_marekebisho_ya_ph(
                ph_ya_sasa=usomaji_wa_sasa['ph'],
                ph_lengwa=ph_lengwa,
                aina_ya_udongo=aina_ya_udongo
            )
            marekebisho['kigezo'] = 'pH ya Udongo'
            marekebisho_yote.append(marekebisho)
        
        elif kigezo in ['N', 'P', 'K']:
            thamani_lengwa = jedwali_la_viwango.loc[zao, (kigezo, 'mean')]
            marekebisho = hesabu_marekebisho_ya_npk(
                kigezo=kigezo,
                thamani_ya_sasa=usomaji_wa_sasa[kigezo],
                thamani_lengwa=thamani_lengwa
            )
            marekebisho_yote.append(marekebisho)
        
        else:
            marekebisho_yote.append({
                'kigezo': kigezo,
                'hatua_inahitajika': None,
                'ujumbe': f'{kigezo} iko nje ya wigo unaofaa, lakini haiwezi kurekebishwa moja kwa moja shambani (inategemea hali ya hewa).'
            })
    
    # MPYA: Angalia pH ya maji, KAMA tu mkulima ameipatia (siyo lazima)
    if ph_ya_maji is not None:
        tathmini_ya_maji = angalia_ph_ya_maji(ph_ya_maji)
        if tathmini_ya_maji['hatua_inahitajika']:
            marekebisho_yote.append(tathmini_ya_maji)
    
    # Kama hakuna tatizo lolote (udongo NA maji vyote viko sawa)
    if len(marekebisho_yote) == 0:
        return {
            'zao': zao,
            'hali_ya_jumla': 'Nzuri',
            'ujumbe': f'Hali ya udongo (na maji, kama yalitolewa) inafaa kwa {zao}. Hakuna marekebisho yanayohitajika.',
            'marekebisho': []
        }
    
    return {
        'zao': zao,
        'hali_ya_jumla': 'Kuna matatizo yanayohitaji hatua',
        'idadi_ya_matatizo': len(marekebisho_yote),
        'marekebisho': marekebisho_yote
    }
```

```python
def angalia_ph_ya_maji(ph_ya_maji, wigo_wa_chini=6.0, wigo_wa_juu=7.5, ujazo_wa_lita=100):
    """
    ph_ya_maji: pH ya sasa ya maji ya umwagiliaji
    wigo_wa_chini/wigo_wa_juu: kiwango cha jumla kinachokubalika
    ujazo_wa_lita: ujazo wa maji unaotibiwa (default lita 100)
    """
    FACTOR_ML_KWA_UNIT = 20   # ml kwa kila lita 100, kwa kila unit ya pH (makadirio ya kawaida)
    
    if ph_ya_maji < wigo_wa_chini:
        tofauti = round(wigo_wa_chini - ph_ya_maji, 2)
        kiasi_ml = round(tofauti * FACTOR_ML_KWA_UNIT * (ujazo_wa_lita / 100), 2)
        return {
            'kigezo': 'pH ya Maji',
            'hatua_inahitajika': True,
            'hali': 'asidi_nyingi',
            'thamani_ya_sasa': ph_ya_maji,
            'wigo_unaofaa': f'{wigo_wa_chini}-{wigo_wa_juu}',
            'aina_ya_marekebisho': 'Potassium Bicarbonate (au Sodium Bicarbonate)',
            'kiasi': kiasi_ml,
            'kipimo': f'ml kwa kila lita {ujazo_wa_lita} za maji',
            'ujumbe': 'Maji yana asidi nyingi. Hii inaweza kufanya Calcium na Magnesium visipatikane vizuri kwa mmea.'
        }
    elif ph_ya_maji > wigo_wa_juu:
        tofauti = round(ph_ya_maji - wigo_wa_juu, 2)
        kiasi_ml = round(tofauti * FACTOR_ML_KWA_UNIT * (ujazo_wa_lita / 100), 2)
        return {
            'kigezo': 'pH ya Maji',
            'hatua_inahitajika': True,
            'hali': 'alkali_nyingi',
            'thamani_ya_sasa': ph_ya_maji,
            'wigo_unaofaa': f'{wigo_wa_chini}-{wigo_wa_juu}',
            'aina_ya_marekebisho': 'Phosphoric Acid (au Citric Acid - salama zaidi)',
            'kiasi': kiasi_ml,
            'kipimo': f'ml kwa kila lita {ujazo_wa_lita} za maji',
            'ujumbe': 'Maji yana alkali nyingi. Hii inaweza kufanya Iron, Manganese, na Zinc visipatikane vizuri kwa mmea.'
        }
    else:
        return {
            'kigezo': 'pH ya Maji',
            'hatua_inahitajika': False,
            'ujumbe': 'pH ya maji ya umwagiliaji iko ndani ya wigo unaofaa.'
        }
```

```python
%%writefile diagnostic_functions.py
"""
Diagnostic Functions - Model B
Kazi hizi zinatumika kutathmini hali ya udongo na maji ya umwagiliaji
kwa zao maalum lililopandwa, na kutoa ushauri wa marekebisho.

Jinsi ya kutumia (kwa backend):
    from diagnostic_functions import toa_ripoti_kamili, pakua_jedwali_la_viwango
    jedwali = pakua_jedwali_la_viwango('viwango_vya_mazao.json')
    ripoti = toa_ripoti_kamili(zao, usomaji_wa_sasa, jedwali, aina_ya_udongo, ph_ya_maji)
"""

import json


def pakua_jedwali_la_viwango(path_ya_json='viwango_vya_mazao.json'):
    """Inapakua JSON na kuirudisha kama dictionary rahisi."""
    with open(path_ya_json) as faili:
        return json.load(faili)


def angalia_hali_ya_udongo(zao, usomaji_wa_sasa, jedwali_la_viwango):
    matatizo = []
    if zao not in jedwali_la_viwango:
        return [f"Samahani, zao '{zao}' halipo kwenye database yetu."]

    for kigezo in ['N', 'P', 'K', 'temperature', 'humidity', 'ph']:
        thamani_ya_sasa = usomaji_wa_sasa[kigezo]
        kiwango_cha_chini = jedwali_la_viwango[zao][kigezo]['min']
        kiwango_cha_juu = jedwali_la_viwango[zao][kigezo]['max']

        if thamani_ya_sasa < kiwango_cha_chini:
            matatizo.append({
                'kigezo': kigezo, 'hali': 'chini_ya_kiwango',
                'thamani_ya_sasa': thamani_ya_sasa,
                'kiwango_kinachohitajika': kiwango_cha_chini,
                'tofauti': round(kiwango_cha_chini - thamani_ya_sasa, 2)
            })
        elif thamani_ya_sasa > kiwango_cha_juu:
            matatizo.append({
                'kigezo': kigezo, 'hali': 'juu_ya_kiwango',
                'thamani_ya_sasa': thamani_ya_sasa,
                'kiwango_kinachohitajika': kiwango_cha_juu,
                'tofauti': round(thamani_ya_sasa - kiwango_cha_juu, 2)
            })
    return matatizo


def hesabu_marekebisho_ya_ph(ph_ya_sasa, ph_lengwa, aina_ya_udongo='tifutifu'):
    buffer_factor = {'mchanga': 1.75, 'tifutifu': 3.5, 'mfinyanzi': 6.5}
    sulfur_factor = {'mchanga': 280, 'tifutifu': 420, 'mfinyanzi': 560}
    tofauti_ya_ph = round(ph_lengwa - ph_ya_sasa, 2)

    if abs(tofauti_ya_ph) < 0.1:
        return {'hatua_inahitajika': False, 'ujumbe': 'pH iko karibu kabisa na kiwango kinachohitajika.'}
    if tofauti_ya_ph > 0:
        kiasi = round(tofauti_ya_ph * buffer_factor[aina_ya_udongo], 2)
        return {'hatua_inahitajika': True, 'aina_ya_marekebisho': 'chokaa (agricultural lime)',
                'kiasi': kiasi, 'kipimo': 'tani kwa hekta',
                'sababu': 'Udongo una asidi nyingi kuliko inavyohitajika kwa zao hili.'}
    else:
        kiasi = round(abs(tofauti_ya_ph) * sulfur_factor[aina_ya_udongo], 2)
        return {'hatua_inahitajika': True, 'aina_ya_marekebisho': 'sulfuri (elemental sulfur)',
                'kiasi': kiasi, 'kipimo': 'kg kwa hekta',
                'sababu': 'Udongo una alkali nyingi kuliko inavyohitajika kwa zao hili.'}


def hesabu_marekebisho_ya_npk(kigezo, thamani_ya_sasa, thamani_lengwa):
    factor = {'N': 2.5, 'P': 2.5, 'K': 2.0}
    majina_kamili = {'N': 'Nitrogen', 'P': 'Phosphorus', 'K': 'Potassium'}
    tofauti = round(thamani_lengwa - thamani_ya_sasa, 2)

    if abs(tofauti) < 2:
        return {'hatua_inahitajika': False,
                'ujumbe': f'{majina_kamili[kigezo]} iko karibu na kiwango kinachohitajika.'}
    if tofauti > 0:
        kiasi = round(tofauti * factor[kigezo], 2)
        return {'hatua_inahitajika': True, 'kigezo': majina_kamili[kigezo], 'hali': 'upungufu',
                'kiasi': kiasi, 'kipimo': 'kg kwa hekta (nutrient safi)',
                'sababu': f'{majina_kamili[kigezo]} imepungua kuliko inavyohitajika kwa zao hili.'}
    else:
        kiasi = round(tofauti * factor[kigezo], 2)
        return {'hatua_inahitajika': True, 'kigezo': majina_kamili[kigezo], 'hali': 'uzidi',
                'kiasi': abs(kiasi), 'kipimo': 'kg kwa hekta (punguza kwa kiasi hiki)',
                'sababu': f'{majina_kamili[kigezo]} imezidi kuliko inavyohitajika kwa zao hili.'}


def angalia_ph_ya_maji(ph_ya_maji, wigo_wa_chini=6.0, wigo_wa_juu=7.5, ujazo_wa_lita=100):
    FACTOR_ML_KWA_UNIT = 20
    if ph_ya_maji < wigo_wa_chini:
        tofauti = round(wigo_wa_chini - ph_ya_maji, 2)
        kiasi_ml = round(tofauti * FACTOR_ML_KWA_UNIT * (ujazo_wa_lita / 100), 2)
        return {'kigezo': 'pH ya Maji', 'hatua_inahitajika': True, 'hali': 'asidi_nyingi',
                'thamani_ya_sasa': ph_ya_maji, 'wigo_unaofaa': f'{wigo_wa_chini}-{wigo_wa_juu}',
                'aina_ya_marekebisho': 'Potassium Bicarbonate (au Sodium Bicarbonate)',
                'kiasi': kiasi_ml, 'kipimo': f'ml kwa kila lita {ujazo_wa_lita} za maji',
                'ujumbe': 'Maji yana asidi nyingi. Hii inaweza kufanya Calcium na Magnesium visipatikane vizuri.'}
    elif ph_ya_maji > wigo_wa_juu:
        tofauti = round(ph_ya_maji - wigo_wa_juu, 2)
        kiasi_ml = round(tofauti * FACTOR_ML_KWA_UNIT * (ujazo_wa_lita / 100), 2)
        return {'kigezo': 'pH ya Maji', 'hatua_inahitajika': True, 'hali': 'alkali_nyingi',
                'thamani_ya_sasa': ph_ya_maji, 'wigo_unaofaa': f'{wigo_wa_chini}-{wigo_wa_juu}',
                'aina_ya_marekebisho': 'Phosphoric Acid (au Citric Acid - salama zaidi)',
                'kiasi': kiasi_ml, 'kipimo': f'ml kwa kila lita {ujazo_wa_lita} za maji',
                'ujumbe': 'Maji yana alkali nyingi. Hii inaweza kufanya Iron, Manganese, Zinc visipatikane vizuri.'}
    else:
        return {'kigezo': 'pH ya Maji', 'hatua_inahitajika': False,
                'ujumbe': 'pH ya maji ya umwagiliaji iko ndani ya wigo unaofaa.'}


def toa_ripoti_kamili(zao, usomaji_wa_sasa, jedwali_la_viwango, aina_ya_udongo='tifutifu', ph_ya_maji=None):
    if zao not in jedwali_la_viwango:
        return {'hitilafu': f"Zao '{zao}' halipo kwenye database yetu."}

    matatizo = angalia_hali_ya_udongo(zao, usomaji_wa_sasa, jedwali_la_viwango)
    marekebisho_yote = []

    for tatizo in matatizo:
        kigezo = tatizo['kigezo']
        if kigezo == 'ph':
            ph_lengwa = jedwali_la_viwango[zao]['ph']['mean']
            marekebisho = hesabu_marekebisho_ya_ph(usomaji_wa_sasa['ph'], ph_lengwa, aina_ya_udongo)
            marekebisho['kigezo'] = 'pH ya Udongo'
            marekebisho_yote.append(marekebisho)
        elif kigezo in ['N', 'P', 'K']:
            thamani_lengwa = jedwali_la_viwango[zao][kigezo]['mean']
            marekebisho = hesabu_marekebisho_ya_npk(kigezo, usomaji_wa_sasa[kigezo], thamani_lengwa)
            marekebisho_yote.append(marekebisho)
        else:
            marekebisho_yote.append({
                'kigezo': kigezo, 'hatua_inahitajika': None,
                'ujumbe': f'{kigezo} iko nje ya wigo unaofaa, lakini haiwezi kurekebishwa moja kwa moja shambani.'
            })

    if ph_ya_maji is not None:
        tathmini_ya_maji = angalia_ph_ya_maji(ph_ya_maji)
        if tathmini_ya_maji['hatua_inahitajika']:
            marekebisho_yote.append(tathmini_ya_maji)

    if len(marekebisho_yote) == 0:
        return {
            'zao': zao, 'hali_ya_jumla': 'Nzuri',
            'ujumbe': f'Hali ya udongo (na maji, kama yalitolewa) inafaa kwa {zao}. Hakuna marekebisho yanayohitajika.',
            'marekebisho': []
        }

    return {
        'zao': zao, 'hali_ya_jumla': 'Kuna matatizo yanayohitaji hatua',
        'idadi_ya_matatizo': len(marekebisho_yote),
        'marekebisho': marekebisho_yote
    }
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X_v4 = df_kamili_v4[['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']]
y_v4 = df_kamili_v4['label']

X_train4, X_test4, y_train4, y_test4 = train_test_split(
    X_v4, y_v4, test_size=0.2, random_state=42, stratify=y_v4
)

model_v4 = RandomForestClassifier(n_estimators=200, random_state=42)
model_v4.fit(X_train4, y_train4)

print("Model A v4 imemaliza kujifunza! Idadi ya mazao:", len(model_v4.classes_))
```

```python
from sklearn.metrics import classification_report, precision_score, recall_score, f1_score

utabiri_v4 = model_v4.predict(X_test4)

print(classification_report(y_test4, utabiri_v4))

print(f"\nPrecision (macro): {precision_score(y_test4, utabiri_v4, average='macro'):.4f}")
print(f"Recall (macro): {recall_score(y_test4, utabiri_v4, average='macro'):.4f}")
print(f"F1-Score (macro): {f1_score(y_test4, utabiri_v4, average='macro'):.4f}")
```
